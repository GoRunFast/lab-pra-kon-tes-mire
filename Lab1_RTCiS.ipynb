## Краткая теория

### АМ-сигнал

Тональный амплитудно-модулированный сигнал имеет вид:

$$
u_{\text{вх}}(t)=U_0\left[1+m\cos(2\pi F_{\text{м}}t)\right]\cos(2\pi f_0t),
$$

где:

- $U_0$ — амплитуда несущего колебания, В;
- $f_0$ — частота несущего колебания, Гц;
- $F_{\text{м}}$ — частота модулирующего сигнала, Гц;
- $m$ — коэффициент глубины модуляции, $0 \le m \le 1$.

После тригонометрического преобразования сигнал можно представить как сумму трёх гармоник:

$$
u_{\text{вх}}(t)=U_0\cos(2\pi f_0t)
+\frac{mU_0}{2}\cos\left[2\pi(f_0-F_{\text{м}})t\right]
+\frac{mU_0}{2}\cos\left[2\pi(f_0+F_{\text{м}})t\right].
$$

Таким образом, в спектре АМ-сигнала присутствуют:

- несущая частота $f_0$ с амплитудой $U_0$;
- нижняя боковая частота $f_0-F_{\text{м}}$ с амплитудой $mU_0/2$;
- верхняя боковая частота $f_0+F_{\text{м}}$ с амплитудой $mU_0/2$.

Ширина спектра тонального АМ-сигнала:

$$
\Delta F_{\text{АМ}}=2F_{\text{м}}.
$$

### Одиночный колебательный контур

Одиночный колебательный контур, настроенный на частоту $f_0$, работает как узкополосный фильтр. Его комплексный коэффициент передачи приближённо описывается выражением:

$$
\dot{K}(jf)=\frac{1}{1+jQ\left(\dfrac{f}{f_0}-\dfrac{f_0}{f}\right)}.
$$

Здесь $Q$ — эквивалентная добротность контура.

Полоса пропускания по уровню $0.707$, то есть на уровне $-3$ дБ, равна:

$$
\Pi_{0.7}=\frac{f_0}{Q}.
$$

Чем больше добротность $Q$, тем уже полоса пропускания. Если полоса контура меньше ширины спектра АМ-сигнала, боковые составляющие ослабляются сильнее несущей. В результате глубина модуляции на выходе уменьшается:

$$
m_{\text{вых}}<m.
$$

## Шаг 1. Подключение библиотек и задание исходных данных

На этом шаге мы:

1. Подключаем библиотеки для численных расчётов и построения графиков.
2. Задаём параметры варианта.
3. Формируем временную сетку для дискретного сигнала.

**Используемые библиотеки:**

- `numpy` — работа с массивами, математические функции, генерация сигналов;
- `matplotlib.pyplot` — построение графиков;
- `scipy.fft` — прямое и обратное дискретное преобразование Фурье.

**Важно:** частоты в таблице указаны в килогерцах, а в расчётах используются герцы. Поэтому значения умножаются на `1e3`.

Частота дискретизации должна быть значительно выше максимальной частоты в спектре сигнала. В данной работе она выбрана равной $200$ кГц, что более чем в два раза превышает верхнюю боковую частоту для всех вариантов.

```python
# Подключаем библиотеку NumPy:
# она позволяет работать с массивами чисел и выполнять математические операции
import numpy as np

# Подключаем библиотеку Matplotlib:
# она используется для построения графиков
import matplotlib.pyplot as plt

# Из SciPy импортируем функции быстрого преобразования Фурье:
# fft — прямое преобразование,
# ifft — обратное преобразование,
# fftfreq — формирование массива частот для спектра
from scipy.fft import fft, ifft, fftfreq

# ============================================================
# ВВЕДИТЕ ПАРАМЕТРЫ СВОЕГО ВАРИАНТА ИЗ ТАБЛИЦЫ
# ============================================================

# Амплитуда несущего колебания, В
amplitude_carrier = 2.0

# Частота несущего колебания, кГц
# В таблице она указана в килогерцах
carrier_frequency_kHz = 20.0

# Частота модулирующего сигнала, кГц
modulation_frequency_kHz = 1.0

# Коэффициент глубины амплитудной модуляции
modulation_index = 0.8

# Добротность первого контура — более широкополосного
quality_factor_1 = 10

# Добротность второго контура — более узкополосного
quality_factor_2 = 40

# ============================================================
# ПЕРЕВОД ЕДИНИЦ ИЗМЕРЕНИЯ
# ============================================================

# Переводим амплитуду несущей в вольты
U0 = amplitude_carrier

# Переводим частоту несущей из килогерц в герцы:
# 1 кГц = 1000 Гц
f0 = carrier_frequency_kHz * 1e3

# Переводим частоту модуляции из килогерц в герцы
Fm = modulation_frequency_kHz * 1e3

# Сохраняем коэффициент модуляции в переменную m
m = modulation_index

# Сохраняем добротности контуров
Q1 = quality_factor_1
Q2 = quality_factor_2

# ============================================================
# ПАРАМЕТРЫ ДИСКРЕТИЗАЦИИ СИГНАЛА
# ============================================================

# Частота дискретизации, Гц
# Она определяет, сколько отсчётов сигнала берётся за одну секунду
sampling_frequency = 200e3

# Длительность анализируемой реализации сигнала, с
signal_duration = 0.01

# Общее число отсчётов сигнала:
# количество отсчётов = частота дискретизации * длительность
number_of_samples = int(sampling_frequency * signal_duration)

# Создаём массив моментов времени от 0 до T
# endpoint=False означает, что последняя точка T не включается,
# чтобы не было повторения первого и последнего отсчёта
time = np.linspace(
    start=0,
    stop=signal_duration,
    num=number_of_samples,
    endpoint=False
)

# ============================================================
# ВЫВОД ИСХОДНЫХ ДАННЫХ ДЛЯ КОНТРОЛЯ
# ============================================================

print('Исходные данные варианта:')
print(f'Амплитуда несущей U0 = {U0:.3f} В')
print(f'Частота несущей f0 = {f0 / 1e3:.3f} кГц')
print(f'Частота модуляции Fm = {Fm / 1e3:.3f} кГц')
print(f'Коэффициент модуляции m = {m:.3f}')
print(f'Добротность первого контура Q1 = {Q1}')
print(f'Добротность второго контура Q2 = {Q2}')
print()
print('Параметры дискретизации:')
print(f'Частота дискретизации = {sampling_frequency / 1e3:.1f} кГц')
print(f'Длительность сигнала = {signal_duration * 1e3:.1f} мс')
print(f'Число отсчётов N = {number_of_samples}')
print(f'Шаг по времени = {signal_duration / number_of_samples * 1e6:.3f} мкс')
```

## Шаг 2. Формирование АМ-сигнала и его огибающей

Теперь сформируем АМ-сигнал по формуле:

$$
u_{\text{вх}}(t)=U_0\left[1+m\cos(2\pi F_{\text{м}}t)\right]\cos(2\pi f_0t).
$$

Огибающая АМ-сигнала описывается выражением:

$$
U_{\text{ог}}(t)=U_0\left[1+m\cos(2\pi F_{\text{м}}t)\right].
$$

Она показывает, как медленно изменяется амплитуда высокочастотного колебания.

Построим график на интервале, равном двум периодам модулирующего сигнала:

$$
t_{\max}=\frac{2}{F_{\text{м}}}.
$$

Также рассчитаем максимальную и минимальную амплитуды огибающей и проверим коэффициент модуляции по формуле:

$$
m_{\text{расч}}=\frac{U_{\max}-U_{\min}}{U_{\max}+U_{\min}}.
$$

```python
# ============================================================
# ФОРМИРОВАНИЕ ОГИБАЮЩЕЙ АМ-СИГНАЛА
# ============================================================

# Огибающая определяет медленное изменение амплитуды сигнала
# Она вычисляется по формуле:
# U_ог(t) = U0 * (1 + m * cos(2 * pi * Fm * t))
envelope = U0 * (1 + m * np.cos(2 * np.pi * Fm * time))

# ============================================================
# ФОРМИРОВАНИЕ ПОЛНОГО АМ-СИГНАЛА
# ============================================================

# АМ-сигнал получается умножением огибающей на несущее колебание:
# u_вх(t) = U_ог(t) * cos(2 * pi * f0 * t)
input_signal = envelope * np.cos(2 * np.pi * f0 * time)

# ============================================================
# ВЫБОР ИНТЕРВАЛА ДЛЯ ГРАФИКА
# ============================================================

# Для наглядности показываем только два периода модулирующего сигнала
# Длительность одного периода модуляции равна 1 / Fm
graph_duration = 2 / Fm

# Создаём логическую маску:
# True означает, что соответствующий момент времени попадает в выбранный интервал
time_mask = time <= graph_duration

# Выбираем только те значения времени и сигнала, которые попали в интервал
time_for_plot = time[time_mask]
signal_for_plot = input_signal[time_mask]
envelope_for_plot = envelope[time_mask]

# ============================================================
# ПОСТРОЕНИЕ ГРАФИКА АМ-СИГНАЛА
# ============================================================

# Создаём новое графическое окно заданного размера
plt.figure(figsize=(12, 5))

# Строим график АМ-сигнала
plt.plot(
    time_for_plot * 1e3,      # переводим время из секунд в миллисекунды
    signal_for_plot,          # значения АМ-сигнала
    color='blue',
    linewidth=1,
    label='АМ-сигнал'
)

# Строим верхнюю огибающую
plt.plot(
    time_for_plot * 1e3,
    envelope_for_plot,
    color='red',
    linestyle='--',
    linewidth=2,
    label='Верхняя огибающая'
)

# Строим нижнюю огибающую
plt.plot(
    time_for_plot * 1e3,
    -envelope_for_plot,
    color='red',
    linestyle='--',
    linewidth=2,
    label='Нижняя огибающая'
)

# Добавляем заголовок графика
plt.title('Осциллограмма входного АМ-сигнала')

# Подписываем ось времени
plt.xlabel('Время, мс')

# Подписываем ось напряжения
plt.ylabel('Напряжение, В')

# Включаем сетку для удобства чтения графика
plt.grid(True)

# Добавляем легенду
plt.legend()

# Автоматически подбираем отступы, чтобы ничего не обрезалось
plt.tight_layout()

# Отображаем график
plt.show()

# ============================================================
# ПРОВЕРКА КОЭФФИЦИЕНТА МОДУЛЯЦИИ
# ============================================================

# Максимальное значение огибающей
U_max = np.max(envelope)

# Минимальное значение огибающей
U_min = np.min(envelope)

# Рассчитываем коэффициент модуляции по амплитудам огибающей
m_calculated = (U_max - U_min) / (U_max + U_min)

# Выводим результаты расчёта
print('Проверка параметров АМ-сигнала:')
print(f'Максимальная амплитуда огибающей U_max = {U_max:.4f} В')
print(f'Минимальная амплитуда огибающей U_min = {U_min:.4f} В')
print(f'Расчётный коэффициент модуляции m_расч = {m_calculated:.4f}')
print(f'Заданный коэффициент модуляции m = {m:.4f}')

# Проверяем, совпадает ли расчётное значение с заданным
if np.isclose(m_calculated, m, atol=1e-6):
    print('Проверка пройдена: расчётный коэффициент модуляции совпадает с заданным.')
else:
    print('Внимание: расчётный коэффициент модуляции не совпадает с заданным.')
```

## Шаг 3. Спектральный анализ входного АМ-сигнала

Для перехода от временного представления сигнала к частотному используется дискретное преобразование Фурье (ДПФ).

В Python его удобно выполнять с помощью функции `fft()` из библиотеки `scipy.fft`.

Амплитудный спектр рассчитывается по формуле:

$$
A(f_k)=\frac{2}{N}\left|\dot{U}(f_k)\right|,
$$

где:

- $N$ — число отсчётов сигнала;
- $\dot{U}(f_k)$ — комплексный спектр сигнала.

Множитель $2/N$ нужен для того, чтобы амплитуда гармоники на графике соответствовала её физической амплитуде в вольтах.

Ожидаемые теоретические амплитуды:

| Составляющая | Частота | Амплитуда |
|---|---:|---:|
| Несущая | $f_0$ | $U_0$ |
| Нижняя боковая | $f_0-F_{\text{м}}$ | $mU_0/2$ |
| Верхняя боковая | $f_0+F_{\text{м}}$ | $mU_0/2$ |

```python
# ============================================================
# ПРЯМОЕ БЫСТРОЕ ПРЕОБРАЗОВАНИЕ ФУРЬЕ
# ============================================================

# Вычисляем комплексный спектр входного АМ-сигнала
input_signal_spectrum = fft(input_signal)

# Формируем массив частот, соответствующих отсчётам спектра
# Первый аргумент — число отсчётов,
# второй аргумент — шаг дискретизации по времени, то есть 1 / fs
frequency_axis = fftfreq(number_of_samples, 1 / sampling_frequency)

# Вычисляем амплитудный спектр:
# сначала берём модуль комплексного спектра,
# затем нормируем его на 2 / N
input_signal_amplitude_spectrum = (2 / number_of_samples) * np.abs(input_signal_spectrum)

# ============================================================
# ВЫБОР ДИАПАЗОНА ЧАСТОТ ДЛЯ ГРАФИКА
# ============================================================

# Нижняя граница диапазона частот:
# берём несущую минус три частоты модуляции
frequency_lower_limit = f0 - 3 * Fm

# Верхняя граница диапазона частот:
# берём несущую плюс три частоты модуляции
frequency_upper_limit = f0 + 3 * Fm

# Создаём маску для выбора только нужного участка спектра
frequency_mask = (
    (frequency_axis >= frequency_lower_limit) &
    (frequency_axis <= frequency_upper_limit)
)

# Выбираем частоты и амплитуды только в нужном диапазоне
frequencies_for_plot = frequency_axis[frequency_mask]
amplitudes_for_plot = input_signal_amplitude_spectrum[frequency_mask]

# ============================================================
# ПОСТРОЕНИЕ АМПЛИТУДНОГО СПЕКТРА
# ============================================================

# Создаём графическое окно
plt.figure(figsize=(12, 5))

# Строим спектр в виде вертикальных линий
plt.stem(
    frequencies_for_plot / 1e3,   # переводим частоту в килогерцы
    amplitudes_for_plot,          # амплитуды гармоник
    basefmt=' '
)

# Добавляем заголовок
plt.title('Амплитудный спектр входного АМ-сигнала')

# Подписываем ось частот
plt.xlabel('Частота, кГц')

# Подписываем ось амплитуды
plt.ylabel('Амплитуда, В')

# Включаем сетку
plt.grid(True)

# Показываем график
plt.tight_layout()
plt.show()

# ============================================================
# АВТОМАТИЧЕСКИЙ ПОИСК АМПЛИТУД СОСТАВЛЯЮЩИХ
# ============================================================

# Функция находит амплитуду спектральной составляющей
# на заданной частоте или максимально близкой к ней
def find_spectrum_amplitude(target_frequency, frequencies, amplitudes):
    # Находим индекс элемента массива частот,
    # который расположен ближе всего к искомой частоте
    nearest_index = np.argmin(np.abs(frequencies - target_frequency))

    # Возвращаем амплитуду в найденной точке спектра
    return amplitudes[nearest_index]

# Ищем амплитуду несущей частоты
carrier_amplitude = find_spectrum_amplitude(f0, frequency_axis, input_signal_amplitude_spectrum)

# Ищем амплитуду нижней боковой составляющей
lower_sideband_amplitude = find_spectrum_amplitude(
    f0 - Fm,
    frequency_axis,
    input_signal_amplitude_spectrum
)

# Ищем амплитуду верхней боковой составляющей
upper_sideband_amplitude = find_spectrum_amplitude(
    f0 + Fm,
    frequency_axis,
    input_signal_amplitude_spectrum
)

# Рассчитываем теоретические амплитуды для сравнения
theoretical_carrier_amplitude = U0
theoretical_sideband_amplitude = m * U0 / 2

# ============================================================
# ВЫВОД РЕЗУЛЬТАТОВ СПЕКТРАЛЬНОГО АНАЛИЗА
# ============================================================

print('Результаты спектрального анализа входного сигнала:')
print()
print(f'Несущая частота: {f0 / 1e3:.3f} кГц')
print(f'Расчётная амплитуда несущей: {carrier_amplitude:.4f} В')
print(f'Теоретическая амплитуда несущей: {theoretical_carrier_amplitude:.4f} В')
print()
print(f'Нижняя боковая частота: {(f0 - Fm) / 1e3:.3f} кГц')
print(f'Расчётная амплитуда нижней боковой: {lower_sideband_amplitude:.4f} В')
print(f'Теоретическая амплитуда боковой: {theoretical_sideband_amplitude:.4f} В')
print()
print(f'Верхняя боковая частота: {(f0 + Fm) / 1e3:.3f} кГц')
print(f'Расчётная амплитуда верхней боковой: {upper_sideband_amplitude:.4f} В')
print(f'Теоретическая амплитуда боковой: {theoretical_sideband_amplitude:.4f} В')
print()
print(f'Ширина спектра АМ-сигнала: {2 * Fm / 1e3:.3f} кГц')
```

## Шаг 4. Расчёт АЧХ колебательных контуров

На этом шаге моделируем два одиночных колебательных контура:

- первый контур имеет меньшую добротность $Q_1$ и более широкую полосу пропускания;
- второй контур имеет большую добротность $Q_2$ и более узкую полосу пропускания.

Комплексный коэффициент передачи контура:

$$
\dot{K}(jf)=\frac{1}{1+jQ\left(\dfrac{f}{f_0}-\dfrac{f_0}{f}\right)}.
$$

Амплитудно-частотная характеристика определяется как модуль комплексного коэффициента передачи:

$$
K(f)=\left|\dot{K}(jf)\right|.
$$

Также рассчитаем полосы пропускания обоих контуров:

$$
\Pi_{0.7}=\frac{f_0}{Q}.
$$

И сравним их с шириной спектра АМ-сигнала:

$$
\Delta F_{\text{АМ}}=2F_{\text{м}}.
$$

```python
# ============================================================
# ФУНКЦИЯ ДЛЯ РАСЧЁТА КОМПЛЕКСНОГО КОЭФФИЦИЕНТА ПЕРЕДАЧИ
# ============================================================

def calculate_contour_transfer_function(frequencies, resonance_frequency, quality_factor):
    \"\"\"
    Функция рассчитывает комплексный коэффициент передачи
    одиночного колебательного контура.

    Параметры:
    frequencies — массив частот, Гц;
    resonance_frequency — резонансная частота контура, Гц;
    quality_factor — добротность контура.

    Возвращает:
    complex_transfer_function — массив комплексных коэффициентов передачи.
    \"\"\"

    # Если в массиве частот встречается ноль,
    # в формуле возникнет деление на ноль.
    # Поэтому заменяем нулевую частоту на очень малое число.
    safe_frequencies = np.where(
        frequencies == 0,
        1e-12,
        frequencies
    )

    # Вычисляем комплексный коэффициент передачи по формуле контура
    complex_transfer_function = 1 / (
        1 + 1j * quality_factor * (
            safe_frequencies / resonance_frequency -
            resonance_frequency / safe_frequencies
        )
    )

    # Возвращаем результат
    return complex_transfer_function

# ============================================================
# РАСЧЁТ КОМПЛЕКСНЫХ КОЭФФИЦИЕНТОВ ПЕРЕДАЧИ
# ============================================================

# Комплексный коэффициент передачи первого контура
transfer_function_1 = calculate_contour_transfer_function(
    frequency_axis,
    f0,
    Q1
)

# Комплексный коэффициент передачи второго контура
transfer_function_2 = calculate_contour_transfer_function(
    frequency_axis,
    f0,
    Q2
)

# Рассчитываем АЧХ первого контура:
# АЧХ равна модулю комплексного коэффициента передачи
frequency_response_1 = np.abs(transfer_function_1)

# Рассчитываем АЧХ второго контура
frequency_response_2 = np.abs(transfer_function_2)

# ============================================================
# ПОСТРОЕНИЕ АЧХ И СПЕКТРА СИГНАЛА
# ============================================================

# Создаём графическое окно
plt.figure(figsize=(12, 5))

# Строим АЧХ первого контура
plt.plot(
    frequencies_for_plot / 1e3,
    frequency_response_1[frequency_mask],
    color='blue',
    linewidth=2,
    label=f'АЧХ контура 1, Q1 = {Q1}'
)

# Строим АЧХ второго контура
plt.plot(
    frequencies_for_plot / 1e3,
    frequency_response_2[frequency_mask],
    color='red',
    linewidth=2,
    label=f'АЧХ контура 2, Q2 = {Q2}'
)

# Проводим горизонтальную линию на уровне 0.707,
# который соответствует ослаблению на 3 дБ
plt.axhline(
    y=0.707,
    color='gray',
    linestyle=':',
    linewidth=1.5,
    label='Уровень 0.707 (-3 дБ)'
)

# Наносим нормированный спектр входного АМ-сигнала
plt.stem(
    frequencies_for_plot / 1e3,
    amplitudes_for_plot / U0,
    linefmt='k--',
    markerfmt='ko',
    basefmt=' ',
    label='Нормированный спектр АМ-сигнала'
)

# Добавляем заголовок
plt.title('АЧХ колебательных контуров и спектр входного АМ-сигнала')

# Подписываем ось частот
plt.xlabel('Частота, кГц')

# Подписываем ось коэффициента передачи
plt.ylabel('Коэффициент передачи')

# Включаем сетку
plt.grid(True)

# Добавляем легенду
plt.legend()

# Показываем график
plt.tight_layout()
plt.show()

# ============================================================
# РАСЧЁТ ПОЛОС ПРОПУСКАНИЯ КОНТУРОВ
# ============================================================

# Полоса пропускания первого контура
bandwidth_1 = f0 / Q1

# Полоса пропускания второго контура
bandwidth_2 = f0 / Q2

# Ширина спектра АМ-сигнала
signal_bandwidth = 2 * Fm

# ============================================================
# ВЫВОД РЕЗУЛЬТАТОВ
# ============================================================

print('Полосы пропускания колебательных контуров:')
print(f'Полоса пропускания контура 1: {bandwidth_1 / 1e3:.3f} кГц')
print(f'Полоса пропускания контура 2: {bandwidth_2 / 1e3:.3f} кГц')
print(f'Ширина спектра АМ-сигнала: {signal_bandwidth / 1e3:.3f} кГц')
print()

# Проверяем, пропускает ли контур весь спектр сигнала без заметного ослабления
if bandwidth_1 >= signal_bandwidth:
    print('Контур 1 имеет полосу пропускания не меньше ширины спектра АМ-сигнала.')
else:
    print('Контур 1 уже, чем спектр АМ-сигнала: возможны частотные искажения.')

if bandwidth_2 >= signal_bandwidth:
    print('Контур 2 имеет полосу пропускания не меньше ширины спектра АМ-сигнала.')
else:
    print('Контур 2 уже, чем спектр АМ-сигнала: боковые составляющие будут ослаблены.')
```

## Шаг 5. Прохождение АМ-сигнала через контуры

Спектральный метод основан на важном свойстве линейных цепей:

$$
\dot{U}_{\text{вых}}(jf)=\dot{U}_{\text{вх}}(jf)\cdot\dot{K}(jf).
$$

То есть спектр выходного сигнала получается умножением спектра входного сигнала на комплексный коэффициент передачи цепи.

После этого с помощью обратного преобразования Фурье `ifft()` можно вернуться во временную область и получить выходной сигнал.

На графике мы не показываем самый первый период модуляции. Это связано с тем, что на границах временной реализации могут возникать краевые эффекты дискретного преобразования Фурье. Поэтому для анализа берём установившийся участок сигнала от $1/F_{\text{м}}$ до $3/F_{\text{м}}$.

```python
# ============================================================
# РАСЧЁТ СПЕКТРОВ ВЫХОДНЫХ СИГНАЛОВ
# ============================================================

# Спектр сигнала после первого контура:
# спектр входа умножается на комплексный коэффициент передачи
output_spectrum_1 = input_signal_spectrum * transfer_function_1

# Спектр сигнала после второго контура
output_spectrum_2 = input_signal_spectrum * transfer_function_2

# ============================================================
# ОБРАТНОЕ ПРЕОБРАЗОВАНИЕ ФУРЬЕ
# ============================================================

# Выполняем обратное преобразование Фурье для первого контура
output_signal_1_complex = ifft(output_spectrum_1)

# Выполняем обратное преобразование Фурье для второго контура
output_signal_2_complex = ifft(output_spectrum_2)

# Из-за погрешностей вычислений результат может содержать очень малую мнимую часть.
# Физический сигнал является вещественным, поэтому берём только действительную часть.
output_signal_1 = np.real(output_signal_1_complex)
output_signal_2 = np.real(output_signal_2_complex)

# ============================================================
# ВЫБОР УСТАНОВИВШЕГОСЯ УЧАСТКА СИГНАЛА
# ============================================================

# Начало участка: пропускаем один период модуляции
steady_state_start = 1 / Fm

# Конец участка: берём три периода модуляции
steady_state_end = 3 / Fm

# Создаём маску для выбора установившегося участка
steady_state_mask = (
    (time >= steady_state_start) &
    (time <= steady_state_end)
)

# Выбираем время и значения сигналов на установившемся участке
time_steady = time[steady_state_mask]
input_signal_steady = input_signal[steady_state_mask]
output_signal_1_steady = output_signal_1[steady_state_mask]
output_signal_2_steady = output_signal_2[steady_state_mask]

# ============================================================
# ПОСТРОЕНИЕ ГРАФИКОВ ВХОДНОГО И ВЫХОДНЫХ СИГНАЛОВ
# ============================================================

# Создаём графическое окно с двумя отдельными графиками
plt.figure(figsize=(12, 8))

# ------------------------------------------------------------
# Первый график: широкий контур
# ------------------------------------------------------------

# Создаём первую область для графика
plt.subplot(2, 1, 1)

# Строим входной сигнал серым цветом для сравнения
plt.plot(
    time_steady * 1e3,
    input_signal_steady,
    color='gray',
    alpha=0.5,
    label='Входной АМ-сигнал'
)

# Строим выходной сигнал первого контура
plt.plot(
    time_steady * 1e3,
    output_signal_1_steady,
    color='blue',
    linewidth=1.2,
    label=f'Выходной сигнал, Q1 = {Q1}'
)

# Добавляем заголовок
plt.title('Прохождение АМ-сигнала через широкополосный контур')

# Подписываем ось напряжения
plt.ylabel('Напряжение, В')

# Включаем сетку
plt.grid(True)

# Добавляем легенду
plt.legend()

# ------------------------------------------------------------
# Второй график: узкий контур
# ------------------------------------------------------------

# Создаём вторую область для графика
plt.subplot(2, 1, 2)

# Строим входной сигнал серым цветом
plt.plot(
    time_steady * 1e3,
    input_signal_steady,
    color='gray',
    alpha=0.5,
    label='Входной АМ-сигнал'
)

# Строим выходной сигнал второго контура
plt.plot(
    time_steady * 1e3,
    output_signal_2_steady,
    color='red',
    linewidth=1.2,
    label=f'Выходной сигнал, Q2 = {Q2}'
)

# Добавляем заголовок
plt.title('Прохождение АМ-сигнала через узкополосный контур')

# Подписываем ось времени
plt.xlabel('Время, мс')

# Подписываем ось напряжения
plt.ylabel('Напряжение, В')

# Включаем сетку
plt.grid(True)

# Добавляем легенду
plt.legend()

# Автоматически корректируем расположение графиков
plt.tight_layout()

# Показываем графики
plt.show()

# ============================================================
# ВЫВОД ПРОМЕЖУТОЧНЫХ РЕЗУЛЬТАТОВ
# ============================================================

print('Сравнение амплитуд входного и выходных сигналов:')
print(f'Максимальное значение входного сигнала: {np.max(np.abs(input_signal_steady)):.4f} В')
print(f'Максимальное значение выходного сигнала контура 1: {np.max(np.abs(output_signal_1_steady)):.4f} В')
print(f'Максимальное значение выходного сигнала контура 2: {np.max(np.abs(output_signal_2_steady)):.4f} В')
```

## Шаг 6. Оценка искажения глубины модуляции

После прохождения через узкополосный контур боковые составляющие могут ослабляться сильнее, чем несущая. Из-за этого уменьшается глубина модуляции выходного сигнала.

Коэффициент модуляции на выходе определяется по амплитудам спектральных составляющих:

$$
m_{\text{вых}}=\frac{A_{\text{ниж}}+A_{\text{верх}}}{A_{\text{нес}}},
$$

где:

- $A_{\text{ниж}}$ — амплитуда нижней боковой составляющей;
- $A_{\text{верх}}$ — амплитуда верхней боковой составляющей;
- $A_{\text{нес}}$ — амплитуда несущей составляющей.

Для симметричного контура, настроенного точно на несущую частоту, амплитуды нижней и верхней боковых составляющих должны быть одинаковыми.

Также рассчитаем относительное уменьшение глубины модуляции:

$$
\Delta m=\frac{m-m_{\text{вых}}}{m}\cdot100\%.
$$

```python
# ============================================================
# РАСЧЁТ АМПЛИТУДНОГО СПЕКТРА ВЫХОДНОГО СИГНАЛА
# ============================================================

# Вычисляем амплитудный спектр сигнала после второго контура
output_signal_2_amplitude_spectrum = (
    2 / number_of_samples
) * np.abs(output_spectrum_2)

# ============================================================
# ПОИСК АМПЛИТУД СОСТАВЛЯЮЩИХ ВЫХОДНОГО СПЕКТРА
# ============================================================

# Находим амплитуду несущей составляющей на выходе
output_carrier_amplitude = find_spectrum_amplitude(
    f0,
    frequency_axis,
    output_signal_2_amplitude_spectrum
)

# Находим амплитуду нижней боковой составляющей на выходе
output_lower_sideband_amplitude = find_spectrum_amplitude(
    f0 - Fm,
    frequency_axis,
    output_signal_2_amplitude_spectrum
)

# Находим амплитуду верхней боковой составляющей на выходе
output_upper_sideband_amplitude = find_spectrum_amplitude(
    f0 + Fm,
    frequency_axis,
    output_signal_2_amplitude_spectrum
)

# ============================================================
# РАСЧЁТ ГЛУБИНЫ МОДУЛЯЦИИ НА ВЫХОДЕ
# ============================================================

# Рассчитываем коэффициент модуляции после второго контура
output_modulation_index = (
    output_lower_sideband_amplitude + output_upper_sideband_amplitude
) / output_carrier_amplitude

# Рассчитываем относительное уменьшение глубины модуляции в процентах
modulation_index_reduction_percent = (
    (m - output_modulation_index) / m * 100
)

# ============================================================
# ВЫВОД РЕЗУЛЬТАТОВ
# ============================================================

print('Результаты анализа выходного сигнала узкополосного контура:')
print()
print(f'Амплитуда несущей на выходе: {output_carrier_amplitude:.4f} В')
print(f'Амплитуда нижней боковой на выходе: {output_lower_sideband_amplitude:.4f} В')
print(f'Амплитуда верхней боковой на выходе: {output_upper_sideband_amplitude:.4f} В')
print()
print(f'Исходный коэффициент модуляции m = {m:.4f}')
print(f'Коэффициент модуляции на выходе m_вых2 = {output_modulation_index:.4f}')
print(f'Уменьшение глубины модуляции: {modulation_index_reduction_percent:.2f} %')
print()

# Строим сравнительную столбчатую диаграмму амплитуд
plt.figure(figsize=(10, 5))

# Названия спектральных составляющих
component_names = [
    'Нижняя боковая',
    'Несущая',
    'Верхняя боковая'
]

# Амплитуды на входе
input_amplitudes = [
    lower_sideband_amplitude,
    carrier_amplitude,
    upper_sideband_amplitude
]

# Амплитуды на выходе второго контура
output_amplitudes = [
    output_lower_sideband_amplitude,
    output_carrier_amplitude,
    output_upper_sideband_amplitude
]

# Позиции столбцов на графике
x_positions = np.arange(len(component_names))

# Ширина одного столбца
bar_width = 0.35

# Строим столбцы для входного спектра
plt.bar(
    x_positions - bar_width / 2,
    input_amplitudes,
    width=bar_width,
    label='Входной сигнал'
)

# Строим столбцы для выходного спектра
plt.bar(
    x_positions + bar_width / 2,
    output_amplitudes,
    width=bar_width,
    label='Выходной сигнал, Q2'
)

# Подписываем ось X названиями составляющих
plt.xticks(
    x_positions,
    component_names
)

# Добавляем заголовок
plt.title('Сравнение спектральных составляющих до и после узкополосного контура')

# Подписываем ось Y
plt.ylabel('Амплитуда, В')

# Включаем сетку
plt.grid(axis='y')

# Добавляем легенду
plt.legend()

# Показываем график
plt.tight_layout()
plt.show()
```

### 1. Почему при увеличении добротности контура глубина модуляции на выходе уменьшается?

**Ответ:**

При увеличении добротности $Q$ полоса пропускания контура уменьшается:

$$
\Pi_{0.7}=\frac{f_0}{Q}.
$$

Поэтому боковые составляющие АМ-сигнала, расположенные на частотах $f_0 \pm F_{\text{м}}$, оказываются дальше от резонансной частоты и сильнее ослабляются. Несущая составляющая при этом ослабляется меньше. В результате отношение амплитуды боковых составляющих к амплитуде несущей уменьшается, а значит, уменьшается и глубина модуляции выходного сигнала.

### 2. Как изменится спектр выходного сигнала при расстройке контура?

**Ответ:**

При расстройке контура его резонансная частота перестаёт совпадать с несущей частотой $f_0$. В этом случае несущая и боковые составляющие попадают на разные участки АЧХ контура.

Если контур настроен на частоту выше несущей, верхняя боковая составляющая может ослабляться меньше, чем нижняя. Если контур настроен ниже несущей, наоборот, сильнее ослабляется верхняя боковая составляющая. В результате амплитуды боковых частот становятся неодинаковыми, а форма огибающей выходного сигнала искажается.

### 3. Какую минимальную полосу пропускания должен иметь контур, чтобы ослабление боковых частот не превышало 3 дБ?

**Ответ:**

Чтобы обе боковые частоты попадали в полосу пропускания на уровне не ниже $0.707$, минимальная полоса пропускания контура должна быть не меньше ширины спектра АМ-сигнала:

$$
\Pi_{0.7}\geq2F_{\text{м}}.
$$

Тогда для заданной несущей частоты максимальная допустимая добротность определяется выражением:

$$
Q_{\max}=\frac{f_0}{2F_{\text{м}}}.
$$

### 4. Почему в Шаге 5 при построении графика мы отступили от начала временной оси?

**Ответ:**

В начале реализации сигнала могут проявляться краевые эффекты дискретного преобразования Фурье. Сигнал рассматривается как периодический, а реальные АМ-сигналы не всегда начинаются и заканчиваются с одинаковой фазой. Поэтому на границах могут возникать небольшие искажения.

## 1. Краткая теория и рабочие формулы

### 1.1. Периодический сигнал

Периодическим называется сигнал, который повторяется через период $T$:
$$s(t) = s(t + T).$$

Основная (первая) частота:
$$f_1 = \frac{1}{T}, \quad \omega_1 = 2\pi f_1.$$

Любой сложный периодический сигнал можно представить суммой простых синусоид — **гармоник** с частотами, кратными основной.

### 1.2. Ряд Фурье (тригонометрическая форма)

$$s(t) = \frac{a_0}{2} + \sum_{n=1}^{\infty}\left[a_n\cos(2\pi n f_1 t) + b_n\sin(2\pi n f_1 t)\right].$$

**Обозначения:**
- $s(t)$ — значение сигнала, В;
- $t$ — время, с (или мс);
- $T$ — период, с; $f_1 = 1/T$ — основная частота, Гц;
- $n$ — номер гармоники ($n = 1, 2, 3, \dots$);
- $a_n, b_n$ — коэффициенты ряда Фурье, В;
- $a_0/2$ — **постоянная составляющая** (среднее значение сигнала), В.

### 1.3. Амплитуда и фаза гармоники

$$A_n = \sqrt{a_n^2 + b_n^2}, \quad \text{В}; \qquad \varphi_n = \operatorname{arctg}\left(-\frac{b_n}{a_n}\right), \quad \text{рад}.$$

Постоянная составляющая:
$$A_0 = \frac{a_0}{2} = \frac{1}{T}\int_0^T s(t)\,dt, \quad \text{В}.$$

### 1.4. Прямоугольная последовательность импульсов

Для последовательности прямоугольных импульсов амплитудой $A$, периодом $T$ и длительностью импульса $\tau$ (скважность $q = T/\tau$):

$$a_0 = \frac{2A}{q}, \qquad a_n = \frac{2A}{n\pi}\sin\left(\frac{n\pi}{q}\right), \qquad b_n = 0.$$

Амплитуда $n$-й гармоники:
$$A_n = \left|\frac{2A}{n\pi}\sin\left(\frac{n\pi}{q}\right)\right|, \quad \text{В}.$$

Для **меандра** ($q = 2$): $A_n = 0$ при чётных $n$; $A_n = \dfrac{2A}{n\pi}$ при нечётных $n$.

### 1.5. Спектр

Спектр периодического сигнала — **линейчатый** (дискретный): гармоники стоят на частотах $f_1, 2f_1, 3f_1, \dots$

**Свойства:**
- при **сдвиге сигнала во времени** амплитудный спектр **не меняется**, а фазовый меняется линейно;
- добавление **постоянного напряжения** $U_0$ меняет только $A_0$;
- чем **короче импульс** (больше $q$), тем **шире** спектр.

## 2. Эталонный пример решения (шаг за шагом)

### Условие

Дан периодический сигнал — последовательность прямоугольных импульсов:
- амплитуда $A = 10$ В;
- период $T = 1$ мс $= 10^{-3}$ с;
- длительность импульса $\tau = 0{,}5$ мс;
- скважность $q = T/\tau = 2$ (меандр).

**Требуется:** найти постоянную составляющую, основную частоту, амплитуды первых четырёх гармоник, описать амплитудный спектр.

### Дано

| Величина | Обозначение | Значение | Единица |
|---|---|---|---|
| Амплитуда | $A$ | 10 | В |
| Период | $T$ | 1 | мс $= 10^{-3}$ с |
| Длительность | $\tau$ | 0,5 | мс |
| Скважность | $q = T/\tau$ | 2 | — |

### Шаг 1. Расчёт основной частоты и периода

$$f_1 = \frac{1}{T} = \frac{1}{10^{-3}\ \text{с}} = 1000\ \text{Гц} = 1\ \text{кГц}.$$
$$\omega_1 = 2\pi f_1 = 2\pi \cdot 1000 \approx 6283\ \text{рад/с}.$$

Период уже задан: $T = 1$ мс.

### Шаг 2. Расчёт постоянной составляющей

$$A_0 = \frac{a_0}{2} = \frac{A}{q} = \frac{10}{2} = 5\ \text{В}.$$

**Физический смысл:** среднее значение меандра с амплитудой 10 В и скважностью 2 равно 5 В.

### Шаг 3. Расчёт амплитуд гармоник

Формула: $A_n = \left|\dfrac{2A}{n\pi}\sin\left(\dfrac{n\pi}{q}\right)\right|$. При $q = 2$:

| $n$ | Частота, кГц | $\sin(n\pi/2)$ | $A_n$, В | Примечание |
|---|---|---|---|---|
| 1 | 1 | 1 | $2\cdot 10/(1\cdot\pi) \approx 6{,}37$ | нечётная |
| 2 | 2 | 0 | 0 | чётная — отсутствует |
| 3 | 3 | −1 | $2\cdot 10/(3\cdot\pi) \approx 2{,}12$ | нечётная |
| 4 | 4 | 0 | 0 | чётная — отсутствует |

Модуль учитывает знак: $|-1| = 1$.

### Шаг 4. Ответ и физический вывод

- Постоянная составляющая $A_0 = 5$ В.
- Основная частота $f_1 = 1$ кГц, период $T = 1$ мс.
- Амплитуды гармоник: $A_1 \approx 6{,}37$ В; $A_2 = 0$; $A_3 \approx 2{,}12$ В; $A_4 = 0$.
- Спектр — линейчатый, с линиями на частотах 1, 3, 5, … кГц. **Чётные гармоники отсутствуют** — характерный признак меандра. Огибающая спектра убывает как $1/n$.


```python

# ============================================================
# ПРОВЕРКА ЭТАЛОННОГО ПРИМЕРА НА PYTHON
# ============================================================
import numpy as np

A = 10.0        # амплитуда, В
T = 1.0e-3      # период, с
tau = 0.5e-3    # длительность, с

f1 = 1.0 / T            # основная частота, Гц
q = T / tau             # скважность
A0 = A / q              # постоянная составляющая, В

n = np.arange(1, 5)     # номера гармоник 1..4
An = np.abs((2*A/(n*np.pi)) * np.sin(n*np.pi/q))   # амплитуды, В

print(f'f1 = {f1:.0f} Гц = {f1/1000:.1f} кГц')
print(f'T = {T*1000:.2f} мс,  q = {q:.2f}')
print(f'A0 = {A0:.2f} В')
for i in range(len(n)):
    print(f'  n={n[i]}: An = {An[i]:.3f} В, частота = {n[i]*f1/1000:.1f} кГц')
```

> 📌 **Обратите внимание:** в каждой задаче сначала идёт условие, затем — полное решение. Студент может использовать его как образец оформления и проверять свои ответы.



### Задача 1. Чтение готового ряда Фурье

**Условие.** Дан ряд Фурье периодического сигнала:
$$s(t) = 4 + 6\cos(2\pi \cdot 500 \cdot t) + 3\cos(2\pi \cdot 1500 \cdot t) + 2\sin(2\pi \cdot 1000 \cdot t), \quad \text{В}.$$

**Требуется определить:**
1. Постоянную составляющую $A_0$, В.
2. Основную частоту $f_1$, Гц, и период $T$, мс.
3. Амплитуды первой, второй и третьей гармоник $A_1, A_2, A_3$, В.
4. Частоты, на которых стоят эти гармоники, Гц.
5. Является ли сигнал чётным, нечётным или общего вида (обосновать).

---

#### Решение шаг за шагом

**Дано:** ряд Фурье вида
$$s(t) = A_0 + a_1\cos(2\pi f_1 t) + a_2\cos(2\pi f_2 t) + b_2\sin(2\pi f_2 t).$$

Сопоставляя с заданным рядом, выписываем коэффициенты:

| Коэффициент | Значение | Единица |
|---|---|---|
| $A_0$ | 4 | В |
| $a_1$ | 6 | В |
| $a_2$ | 3 | В |
| $b_2$ | 2 | В |
| $f_1$ | 500 | Гц |
| $f_2$ | 1000 | Гц |

**Шаг 1. Постоянная составляющая.**
$$A_0 = 4\ \text{В}.$$
Это среднее значение сигнала за период.

**Шаг 2. Основная частота и период.**
Основная частота — **наименьшая** из присутствующих: $f_1 = 500$ Гц.
$$T = \frac{1}{f_1} = \frac{1}{500} = 2\ \text{мс} = 2\cdot 10^{-3}\ \text{с}.$$

**Шаг 3. Амплитуды гармоник.**
- Первая гармоника ($n=1$, частота $f_1 = 500$ Гц): только косинус, $a_1 = 6$ В, $b_1 = 0$.
  $$A_1 = \sqrt{a_1^2 + b_1^2} = 6\ \text{В}.$$
- Вторая гармоника ($n=2$, частота $f_2 = 1000$ Гц): есть и косинус, и синус.
  $$A_2 = \sqrt{a_2^2 + b_2^2} = \sqrt{3^2 + 2^2} = \sqrt{13} \approx 3{,}61\ \text{В}.$$
- Третья гармоника ($n=3$, частота $1500$ Гц): в ряду **отсутствует** слагаемое с этой частотой.
  $$A_3 = 0\ \text{В}.$$

**Шаг 4. Частоты гармоник.**
Гармоники стоят на частотах, кратных основной:
- 1-я: $f_1 = 500$ Гц;
- 2-я: $f_2 = 1000$ Гц;
- 3-я: $f_3 = 1500$ Гц (амплитуда нулевая).

**Шаг 5. Чётность сигнала.**
- Сигнал **чётный**, если $s(-t) = s(t)$ — в ряду только косинусы ($b_n = 0$).
- Сигнал **нечётный**, если $s(-t) = -s(t)$ — в ряду только синусы ($a_n = 0$, $A_0 = 0$).
- В нашем ряду есть **и косинусы, и синус** ($b_2 = 2 \neq 0$), значит сигнал **общего вида** (ни чётный, ни нечётный).

**Ответ:**

| Величина | Значение |
|---|---|
| $A_0$ | 4 В |
| $f_1$ | 500 Гц |
| $T$ | 2 мс |
| $A_1$ | 6 В |
| $A_2$ | $\sqrt{13} \approx 3{,}61$ В |
| $A_3$ | 0 В |
| Чётность | общего вида (есть и косинусы, и синус) |

**Физический вывод:** сигнал содержит постоянную составляющую 4 В, первую гармонику 500 Гц с амплитудой 6 В и вторую гармонику 1000 Гц с амплитудой 3,61 В. Третья гармоника отсутствует. Наличие синусной составляющей говорит о том, что сигнал не симметричен относительно оси ординат.


```python

# ============================================================
# ЗАДАЧА 1. ПРОВЕРКА НА PYTHON
# ============================================================
import numpy as np

# Коэффициенты из условия
A0 = 4.0      # постоянная составляющая, В
a1 = 6.0      # коэффициент при cos(2*pi*500*t), В
a2 = 3.0      # коэффициент при cos(2*pi*1000*t), В
b2 = 2.0      # коэффициент при sin(2*pi*1000*t), В
f1 = 500.0    # основная частота, Гц
f2 = 1000.0   # частота второй гармоники, Гц

# Расчёт
T = 1.0 / f1                              # период, с
A1 = np.sqrt(a1**2 + 0**2)                # амплитуда 1-й гармоники, В
A2 = np.sqrt(a2**2 + b2**2)               # амплитуда 2-й гармоники, В
A3 = 0.0                                  # 3-й гармоники нет

print(f'A0 = {A0:.2f} В')
print(f'f1 = {f1:.0f} Гц,  T = {T*1000:.2f} мс')
print(f'A1 = {A1:.2f} В (частота {f1:.0f} Гц)')
print(f'A2 = {A2:.3f} В (частота {f2:.0f} Гц)')
print(f'A3 = {A3:.2f} В (частота {3*f1:.0f} Гц)')
print('Чётность: общего вида (b2 != 0)')
```


### Задача 2. Расчёт спектра прямоугольного сигнала

**Условие.** Дана последовательность прямоугольных импульсов:
- амплитуда $A = 8$ В;
- частота $f_1 = 2$ кГц;
- длительность импульса $\tau = 0{,}125$ мс.

**Требуется:**
1. Найти период $T$, мс, и скважность $q$.
2. Вычислить постоянную составляющую $A_0$, В.
3. Рассчитать амплитуды первых пяти гармоник $A_1 \dots A_5$, В.
4. Указать, какие гармоники равны нулю и почему.
5. Описать амплитудный спектр (на каких частотах стоят линии).

---

#### Решение шаг за шагом

**Дано:**

| Величина | Обозначение | Значение | Единица |
|---|---|---|---|
| Амплитуда | $A$ | 8 | В |
| Частота | $f_1$ | 2 | кГц $= 2000$ Гц |
| Длительность | $\tau$ | 0,125 | мс $= 0{,}125\cdot 10^{-3}$ с |

**Шаг 1. Период и скважность.**
$$T = \frac{1}{f_1} = \frac{1}{2000} = 0{,}5\cdot 10^{-3}\ \text{с} = 0{,}5\ \text{мс}.$$
$$q = \frac{T}{\tau} = \frac{0{,}5\ \text{мс}}{0{,}125\ \text{мс}} = 4.$$

**Шаг 2. Постоянная составляющая.**
$$A_0 = \frac{A}{q} = \frac{8}{4} = 2\ \text{В}.$$
Физический смысл: среднее значение сигнала за период — 2 В.

**Шаг 3. Амплитуды гармоник.**
Формула: $A_n = \left|\dfrac{2A}{n\pi}\sin\left(\dfrac{n\pi}{q}\right)\right|$, при $q = 4$:

| $n$ | Частота, кГц | $\sin(n\pi/4)$ | $A_n$, В | Примечание |
|---|---|---|---|---|
| 1 | 2 | $\sqrt{2}/2 \approx 0{,}707$ | $16/(\pi)\cdot 0{,}707 \approx 3{,}60$ | — |
| 2 | 4 | $1$ | $16/(2\pi)\cdot 1 \approx 2{,}55$ | — |
| 3 | 6 | $\sqrt{2}/2 \approx 0{,}707$ | $16/(3\pi)\cdot 0{,}707 \approx 1{,}20$ | — |
| 4 | 8 | $0$ | $0$ | **нулевая** (каждая 4-я) |
| 5 | 10 | $-\sqrt{2}/2 \approx -0{,}707$ | $16/(5\pi)\cdot 0{,}707 \approx 0{,}72$ | модуль |

**Шаг 4. Нулевые гармоники.**
При $q = 4$ нулевыми будут гармоники с номерами, кратными 4: $n = 4, 8, 12, \dots$
Причина: $\sin(n\pi/4) = 0$ при $n = 4k$.

**Шаг 5. Описание спектра.**
Спектр линейчатый, линии стоят на частотах $2, 4, 6, 8, 10, \dots$ кГц. Каждая 4-я линия ($8, 16, \dots$ кГц) отсутствует. Огибающая убывает как $1/n$.

**Ответ:**

| Величина | Значение |
|---|---|
| $T$ | 0,5 мс |
| $q$ | 4 |
| $A_0$ | 2 В |
| $A_1$ | ≈ 3,60 В |
| $A_2$ | ≈ 2,55 В |
| $A_3$ | ≈ 1,20 В |
| $A_4$ | 0 В (кратная 4) |
| $A_5$ | ≈ 0,72 В |

**Физический вывод:** при скважности $q = 4$ постоянная составляющая равна $A/4 = 2$ В, а в спектре «проваливаются» каждая 4-я гармоника. Это объясняется тем, что $\sin(n\pi/4) = 0$ при $n = 4k$.


```python

# ============================================================
# ЗАДАЧА 2. ПРОВЕРКА НА PYTHON
# ============================================================
import numpy as np

A = 8.0            # амплитуда, В
f1 = 2000.0        # частота, Гц
tau = 0.125e-3     # длительность, с

T = 1.0 / f1       # период, с
q = T / tau        # скважность
A0 = A / q         # постоянная составляющая, В

n = np.arange(1, 6)                               # номера гармоник 1..5
An = np.abs((2*A/(n*np.pi)) * np.sin(n*np.pi/q))  # амплитуды, В

print(f'T = {T*1000:.3f} мс,  q = {q:.2f}')
print(f'A0 = {A0:.2f} В')
print('n   | частота, кГц | An, В')
print('----+--------------+-------')
for i in range(len(n)):
    print(f'{n[i]:2d}  | {n[i]*f1/1000:12.1f} | {An[i]:6.3f}')
print()
print('Нулевые гармоники (n кратные 4):', [i for i in n if abs(An[i-1]) < 1e-9])
```


### Задача 3. Качественный вопрос на понимание свойств спектра

**Условие.** Имеется периодический сигнал $s(t)$ с известным амплитудным и фазовым спектром. Сигнал **задержали во времени** на $\Delta t$ (то есть рассмотрели $s(t - \Delta t)$) и одновременно **добавили постоянное напряжение** $U_0$.

**Ответьте:**
1. Что произойдёт с амплитудным спектром при задержке во времени? Изменится ли он?
2. Что произойдёт с фазовым спектром при задержке?
3. Как добавление постоянного напряжения $U_0$ отразится на амплитудном спектре?
4. Изменится ли при этом основная частота сигнала?

---

#### Решение шаг за шагом

**Шаг 1. Задержка во времени — амплитудный спектр.**
Ряд Фурье сигнала $s(t)$:
$$s(t) = A_0 + \sum_n A_n\cos(2\pi n f_1 t + \varphi_n).$$
После задержки $s(t - \Delta t)$:
$$s(t - \Delta t) = A_0 + \sum_n A_n\cos\big(2\pi n f_1 (t - \Delta t) + \varphi_n\big) = A_0 + \sum_n A_n\cos\big(2\pi n f_1 t + \varphi_n - 2\pi n f_1 \Delta t\big).$$
**Амплитуды $A_n$ не изменились.** Значит, **амплитудный спектр не меняется**.

**Шаг 2. Задержка во времени — фазовый спектр.**
Из той же формулы видно, что к каждой начальной фазе добавилось слагаемое:
$$\Delta\varphi_n = -2\pi n f_1 \Delta t.$$
**Фазовый спектр меняется линейно** по номеру гармоники $n$ (или по частоте $f = nf_1$). Коэффициент наклона пропорционален $\Delta t$.

**Шаг 3. Добавление постоянного напряжения $U_0$.**
Постоянное напряжение — это нулевая гармоника (частота 0 Гц). Оно просто прибавляется к постоянной составляющей:
$$A_0 \to A_0 + U_0.$$
Все остальные гармоники ($A_1, A_2, \dots$) **не меняются**.

**Шаг 4. Основная частота.**
Ни задержка во времени, ни добавление постоянного напряжения **не меняют период** $T$ и основную частоту $f_1 = 1/T$. Сигнал остаётся периодическим с тем же периодом.

**Ответ:**

| Вопрос | Ответ |
|---|---|
| 1. Амплитудный спектр при задержке | **Не меняется** ($|A_n|$ те же) |
| 2. Фазовый спектр при задержке | Меняется линейно: $\Delta\varphi_n = -2\pi n f_1\Delta t$ |
| 3. Влияние $U_0$ на амплитудный спектр | Меняется только $A_0 \to A_0 + U_0$ |
| 4. Основная частота | **Не меняется** |

**Физический вывод:** задержка сигнала во времени — это «сдвиг фазы» для каждой гармоники; амплитуды при этом сохраняются. Постоянное напряжение сдвигает весь сигнал вверх, не затрагивая его переменную часть.

**Шаг 4.** Создадим массив номеров гармоник $n = 1, 2, 3, 4, 5$.
**Шаг 5.** По формуле $A_n = \left|\dfrac{2A}{n\pi}\sin\left(\dfrac{n\pi}{q}\right)\right|$ вычислим амплитуды.
**Шаг 6.** Выведем $A_0$ и $A_n$ на печать.
